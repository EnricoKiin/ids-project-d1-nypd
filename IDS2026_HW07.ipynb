### Dataset 1 -- NYPD Arrests Data (Historic)
The dataset was first published in 2018 by the New York City Police Department (NYPD). It contains historical records of arrests made in NYC, dating back to 2006.

The data originates from NYPD administrative records documenting arrests made by police officers. Before publication, the data is reviewed by the NYPD's Office of Management Analysis and Planning. The dataset is updated periodically to reflect corrections and changes to existing records, while additional historical data is released annually.

The dataset was made publicly available to promote transparency and allow researchers, policymakers, and residents to examine arrest patterns across NYC.

Importantly, the dataset represents recorded arrests rather than all crimes committed. An arrest does not necessarily mean that the arrested individual was guilty of the alleged offense.

Source: Footnote and Data dictionary of [NYPD Arrest Data](https://data.cityofnewyork.us/Public-Safety/NYPD-Arrests-Data-Historic-/8h9b-rp9u/about_data)


### Dataset 2 -- NYPD Complaint Data Historic
The dataset was first published in 2016 by the NYPD. It contains historical records of felony, misdemeanor, and violation complaints reported to the NYPD.

The data originates from official police complaint records and is reviewed by the Office of Management Analysis and Planning before publication. The dataset is periodically updated to incorporate corrections, while new historical records are added annually.

Its purpose is to provide public access to reported crime information, allowing researchers, policymakers, and residents to investigate crime trends, geographical patterns, and changes over time.

However, the dataset only captures incidents recorded by the NYPD. Crimes that are never reported or recorded are not represented.

Source: Footnote and Data Dictionary of [NYPD Complaint Data](https://data.cityofnewyork.us/Public-Safety/NYPD-Complaint-Data-Historic/qgea-i56i/about_data)

Arrest numbers are influenced by policing practices, resource allocation, and enforcement priorities. Areas with higher arrest counts are not necessarily more dangerous. Using historical arrest data without considering these factors could reinforce existing biases.

```python
sns.set_theme(style="whitegrid")

nypdComplaintDF["CMPLNT_FR"] = pd.to_datetime(
    nypdComplaintDF["CMPLNT_FR"], errors="coerce"
)

yearly_crimes = (
    nypdComplaintDF
    .loc[nypdComplaintDF["CMPLNT_FR"].dt.year.between(2006, 2025)]
    .groupby(nypdComplaintDF["CMPLNT_FR"].dt.year)
    .size()
)

plt.figure(figsize=(12, 5))
sns.lineplot(x=yearly_crimes.index, y=yearly_crimes.values, marker="o")

plt.title("Annual Reported Crimes in NYC (2006–2025)")
plt.xlabel("Year")
plt.ylabel("Number of Reported Crimes")
plt.xticks(range(2006, 2026, 2))
plt.tight_layout()
plt.show()
```

```python
monthly_data = nypdComplaintDF.loc[
    nypdComplaintDF["CMPLNT_FR"].dt.year.between(2010, 2024),
    ["CMPLNT_FR"]
].copy()

monthly_data["year"] = monthly_data["CMPLNT_FR"].dt.year
monthly_data["month"] = monthly_data["CMPLNT_FR"].dt.month

monthly_counts = (
    monthly_data
    .groupby(["year", "month"])
    .size()
    .unstack(fill_value=0)
)

monthly_average = monthly_counts.mean(axis=0)

plt.figure(figsize=(11, 5))
sns.barplot(x=monthly_average.index, y=monthly_average.values)

plt.title("Average Monthly Reported Crimes in NYC (2010–2024)")
plt.xlabel("Month")
plt.ylabel("Average Number of Reports")
plt.tight_layout()
plt.show()
```

```python
top_crimes = (
    nypdComplaintDF["OFNS_DESC"]
    .value_counts()
    .head(10)
)

plt.figure(figsize=(12, 6))

plt.barh(
    top_crimes.index.astype(str),
    top_crimes.values
)

plt.gca().invert_yaxis()

plt.title("Top 10 Most Frequently Reported Offense Categories")
plt.xlabel("Number of Reports")
plt.ylabel("Offense Category")
plt.tight_layout()
plt.show()
```

```python
borough_names = {
    "B": "Bronx",
    "K": "Brooklyn",
    "M": "Manhattan",
    "Q": "Queens",
    "S": "Staten Island"
}

arrest_boroughs = (
    nypdArrestDF["ARREST_BORO"]
    .astype("string")
    .map(borough_names)
)

complaint_boroughs = (
    nypdComplaintDF["BORO_NM"]
    .astype("string")
    .str.strip()
    .str.title()
)

arrest_counts = arrest_boroughs.value_counts()
complaint_counts = complaint_boroughs.value_counts()

comparison = pd.DataFrame({
    "Arrests": arrest_counts,
    "Complaints": complaint_counts
}).fillna(0)

comparison = comparison.reindex(
    borough_names.values(),
    fill_value=0
)

comparison_pct = comparison.div(
    comparison.sum(axis=0),
    axis=1
) * 100

comparison_pct.plot(
    kind="bar",
    figsize=(11, 5)
)

plt.title("Geographical Distribution of Arrests and Complaints")
plt.xlabel("Borough")
plt.ylabel("Percentage of Dataset Total")
plt.xticks(rotation=0)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
geo_df = nypdComplaintDF[
    ["Latitude", "Longitude"]
].copy()

geo_df = geo_df.dropna()

# Approximate NYC geographical bounds.
geo_df = geo_df[
    geo_df["Latitude"].between(40.45, 40.95) &
    geo_df["Longitude"].between(-74.30, -73.65)
]

plt.figure(figsize=(10, 9))

plt.hexbin(
    geo_df["Longitude"],
    geo_df["Latitude"],
    gridsize=85,
    bins="log",
    mincnt=1,
    cmap="YlOrRd"
)

plt.colorbar(label="Reported Crime Density (log scale)")

plt.title("Geographical Density of Reported Crimes in NYC")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.axis("equal")
plt.tight_layout()
plt.show()
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 160)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

LINE = "═" * 78
SUB  = "─" * 78

def header(title):
    print(f"\n{LINE}\n  {title}\n{LINE}")

def sub(title):
    print(f"\n{SUB}\n  {title}\n{SUB}")

def kv(label, value, width=42):
    print(f"  {label:<{width}} {value}")

# Helper functions
def missing_table(df, name):
    m = df.isna().sum()
    m = m[m > 0].sort_values(ascending=False)
    if m.empty:
        header(f"{name}: missing values")
        print("  (none)")
        return
    out = pd.DataFrame({
        "missing": m.astype(int),
        "pct":     (m / len(df) * 100).round(2),
    })
    header(f"{name}: missing values - {len(out)} of {df.shape[1]} columns have NaN")
    display(out)

def sentinel_table(s, name):
    vc = s.astype("string").value_counts(dropna=False)
    top = vc.head(15)
    out = pd.DataFrame({
        "count": top.values,
        "pct":   (top.values / len(s) * 100).round(2),
    }, index=top.index)
    out.index.name = "value"
    sub(f"{name}: top values (NaN shown explicitly)")
    display(out)

# 1. MISSING
missing_table(nypdArrestDF,    "Arrests")
missing_table(nypdComplaintDF, "Complaints")

# 2. DUPLICATES
header("Duplicates")
kv("Arrest - ARREST_KEY duplicated rows",    f"{nypdArrestDF['ARREST_KEY'].duplicated().sum():,}")
kv("Arrest - full-row duplicates",           f"{nypdArrestDF.duplicated().sum():,}")
kv("Complaint - CMPLNT_NUM duplicated rows",  f"{nypdComplaintDF['CMPLNT_NUM'].duplicated().sum():,}")
kv("Complaint - full-row duplicates",         f"{nypdComplaintDF.duplicated().sum():,}")

# 3. IMPOSSIBLE / INCONSISTENT
header("Impossible / inconsistent values")

sub("Coordinate ranges")
for df, name in [(nypdArrestDF, "Arrests"), (nypdComplaintDF, "Complaints")]:
    lat, lon = df["Latitude"], df["Longitude"]
    n_total = len(df)
    n_have  = (lat.notna() & lon.notna()).sum()
    print(f"\n  {name}")
    kv("  lat range", f"{lat.min():.4f} -> {lat.max():.4f}")
    kv("  lon range", f"{lon.min():.4f} -> {lon.max():.4f}")
    kv("  rows with both coords", f"{n_have:,}  ({n_have/n_total*100:.2f}%)")

sub("Date consistency")
kv("Arrest date range", f"{nypdArrestDF['ARREST_DATE'].min().date()}  ->  {nypdArrestDF['ARREST_DATE'].max().date()}")
kv("Complaint CMPLNT_FR range", f"{nypdComplaintDF['CMPLNT_FR'].min()}  ->  {nypdComplaintDF['CMPLNT_FR'].max()}")
kv("CMPLNT_TO < CMPLNT_FR (end before start)",
   f"{(nypdComplaintDF['CMPLNT_TO'] < nypdComplaintDF['CMPLNT_FR']).sum():,}")
kv("RPT_DT < CMPLNT_FR (reported before occurred)",
   f"{(nypdComplaintDF['RPT_DT'] < nypdComplaintDF['CMPLNT_FR']).sum():,}")
kv("Arrest dates in the future",
   f"{(nypdArrestDF['ARREST_DATE'] > pd.Timestamp.today()).sum():,}")

# 4. SENTINELS
header("Sentinels / coded 'unknown' values in categoricals")
for col in ["SUSP_AGE_GROUP","VIC_AGE_GROUP","SUSP_RACE","VIC_RACE","SUSP_SEX","VIC_SEX"]:
    sentinel_table(nypdComplaintDF[col], f"Complaints, {col}")
for col in ["AGE_GROUP","PERP_RACE","PERP_SEX"]:
    sentinel_table(nypdArrestDF[col], f"Arrests, {col}")

# 5. WEEKLY TARGET DESCRIPTION
header("Weekly complaint counts - descriptive statistics")
weekly = (nypdComplaintDF
          .dropna(subset=["CMPLNT_FR"])
          .loc[nypdComplaintDF["CMPLNT_FR"].dt.year.between(2006, 2025)]
          .set_index("CMPLNT_FR").resample("W").size())
desc = weekly.describe().to_frame().T
desc.columns = [c.capitalize() for c in desc.columns]
display(desc)
kv("Min week", f"{weekly.idxmin().date()}   ({weekly.min():,})")
kv("Max week", f"{weekly.idxmax().date()}   ({weekly.max():,})")
```
