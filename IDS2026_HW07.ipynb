**Answer:**
### Dataset 1 -- NYPD Arrests Data (Historic)
It was created in 2018 by the NYPD (New York Police Department) in conjunction with the Office of Management Analysis and Planning who review the data before publishing. Data was made available to all to allow the public to explore what crimes are being committed in NYC and that have been effected by the NYPD (i.e arrests have been made). Data is updated quarterly as sometimes an arrest may be voided due to the arrestee being found innocent and new rows are added annualy. The data includes all still valid arrests from 2006 till 2025 as of 11th November 2026. (Source: Footnote and Data dictionary of [NYPD Arrest Data](https://data.cityofnewyork.us/Public-Safety/NYPD-Arrests-Data-Historic-/8h9b-rp9u/about_data))


### Dataset 2 -- NYPD Complaint Data Historic
It was created in 2016 by the NYPD and includes all valid felony, misdemeanor, and violation crimes reported to the NYPD. This data is also updated quartely with the help of the Office of Management Analysis and Planning who review it before publishing but new data is added annualy. The data was made available to the masses to allow us to explore what sorts of crime happens in NYC. (Source: Footnote and Data Dictionary of [NYPD Complaint Data](https://data.cityofnewyork.us/Public-Safety/NYPD-Complaint-Data-Historic/qgea-i56i/about_data).)
